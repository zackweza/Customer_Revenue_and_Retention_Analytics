**Objective:** Assess the proportion of total customer revenue generated by the ten highest-revenue customers.

**Method:** SQL was used to aggregate revenue by customer, rank customers by revenue, and calculate the combined revenue and percentage contribution of the top ten.

```python

query = """
WITH customer_revenue AS (
    SELECT
        CustomerID,
        SUM(Quantity * UnitPrice) AS total_revenue
    FROM sales
    WHERE CustomerID IS NOT NULL
      AND InvoiceNo NOT LIKE 'C%'
    GROUP BY CustomerID
),
ranked_customers AS (
    SELECT
        CustomerID,
        total_revenue,
        ROW_NUMBER() OVER (
            ORDER BY total_revenue DESC
        ) AS customer_rank
    FROM customer_revenue
)
SELECT
    ROUND(SUM(
        CASE WHEN customer_rank <= 10
             THEN total_revenue ELSE 0 END
    ), 2) AS top_10_revenue,
    ROUND(SUM(total_revenue), 2) AS total_customer_revenue,
    ROUND(
        100.0 * SUM(
            CASE WHEN customer_rank <= 10
                 THEN total_revenue ELSE 0 END
        ) / SUM(total_revenue), 2
    ) AS top_10_revenue_percentage
FROM ranked_customers;
"""

result = pd.read_sql_query(query, conn)
print(result.to_string(index=False))
```

### Business Question 7: Do Customers Make Repeat Purchases?

**Objective:** Determine how many customers made purchases across multiple invoices compared with customers who purchased only once.

**Method:** SQL was used to count distinct non-cancelled invoices for each identified customer. Customers with more than one invoice were classified as repeat customers, while those with exactly one invoice were classified as one-time customers.

```python

query = """
WITH customer_invoices AS (
    SELECT
        CustomerID,
        COUNT(DISTINCT InvoiceNo) AS total_invoices
    FROM sales
    WHERE CustomerID IS NOT NULL
      AND InvoiceNo NOT LIKE 'C%'
    GROUP BY CustomerID
)
SELECT
    COUNT(*) AS total_customers,
    SUM(
        CASE WHEN total_invoices > 1
             THEN 1 ELSE 0 END
    ) AS repeat_customers,
    SUM(
        CASE WHEN total_invoices = 1
             THEN 1 ELSE 0 END
    ) AS one_time_customers,
    ROUND(
        100.0 * SUM(
            CASE WHEN total_invoices > 1
                 THEN 1 ELSE 0 END
        ) / COUNT(*), 2
    ) AS repeat_customer_percentage
FROM customer_invoices;
"""

result = pd.read_sql_query(query, conn)
print(result.to_string(index=False))
```

**Key Findings:**
- Total customers analyzed: 4,339.
- Repeat customers: 2,845 (65.57%).
- One-time customers: 1,494 (34.43%).

**Business Interpretation:** Most identified customers made purchases across multiple invoices during the observed period. However, approximately one-third purchased only once. The business could investigate why these customers did not return and evaluate appropriate strategies to encourage subsequent purchases.

**Limitation:** Repeat purchasing across the dataset period does not establish customer retention within a specific time interval or predict future purchasing behavior.

```python

query = """
WITH invoice_revenue AS (
    SELECT
        Country,
        InvoiceNo,
        SUM(Quantity * UnitPrice) AS invoice_value
    FROM sales
    WHERE InvoiceNo NOT LIKE 'C%'
    GROUP BY Country, InvoiceNo
)
SELECT
    Country,
    COUNT(*) AS invoice_count,
    ROUND(AVG(invoice_value), 2) AS average_order_value
FROM invoice_revenue
GROUP BY Country
HAVING COUNT(*) >= 10
ORDER BY average_order_value DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
print(result.to_string(index=False))
```

## Business Question 12: Which Countries Have the Highest Cancellation Rates?

**Business Question:** Which countries have the highest cancellation rates?

**Analytical Approach:** Use SQL to count cancelled invoices and total invoices for each country, then calculate the cancellation rate as a percentage. Include countries with at least 10 invoices to reduce the influence of very small transaction counts.

```python
query = """
SELECT
    Country,
    COUNT(DISTINCT CASE
        WHEN InvoiceNo LIKE 'C%' THEN InvoiceNo
    END) AS cancelled_invoices,
    COUNT(DISTINCT InvoiceNo) AS total_invoices,
    ROUND(
        100.0 * COUNT(DISTINCT CASE
            WHEN InvoiceNo LIKE 'C%' THEN InvoiceNo
        END)
        / COUNT(DISTINCT InvoiceNo),
        2
    ) AS cancellation_rate_pct
FROM sales
GROUP BY Country
HAVING COUNT(DISTINCT InvoiceNo) >= 10
ORDER BY cancellation_rate_pct DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)
print(result.to_string(index=False))
```
